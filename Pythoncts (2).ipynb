```python
# 38 method Chaining
class Employee:
    def __init__(self):
        self.salary=0
    def set_salary(self,s):
        if s>0:
            self.salary=s
        return self
    def add_raise(self,r):
        if r >0:
            self.salary=self.salary+r
        return self
    def display(self):
        print("Final Salary:",self.salary)
e=Employee()
e.set_salary(20000).add_raise(5000).display()
```

```python
# 40. Class Methods
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    @classmethod
    def from_string(cls, data):
        name, sal = data.split(",")
        return cls(name, int(sal))
    def display(self):
        print("Name :", self.name)
        print("Salary :", self.salary)
s = "MADES,75000"
e = Employee.from_string(s)
e.display()
```

```python
def calculate(a,b,op):
    if op=="+":
        return a+b
    elif op=="-":
        return a-b
    elif op=="*":
        return a*b
    elif op=="/":
        if b!=0:
            return a/b
        else:
            return "Cannot divide by zero"

a=float(input())
b=float(input())
op=input()

print(calculate(a,b,op))
```
